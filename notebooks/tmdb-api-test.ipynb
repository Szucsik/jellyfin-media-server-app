```python
import requests

API_KEY = "3a0082cf8d011814183d747aa172d999"
IMDB_ID = "tt0283226"  # Example: The Shawshank Redemption

url = f"https://api.themoviedb.org/3/find/{IMDB_ID}"
params = {
    "api_key": API_KEY,
    "external_source": "imdb_id"
}

response = requests.get(url, params=params)
data = response.json()

# Check movies first, then TV shows
if data["movie_results"]:
    title = data["movie_results"][0]["title"]
    print(f"Movie: {title}")
elif data["tv_results"]:
    title = data["tv_results"][0]["name"]
    print(f"TV Show: {title}")
else:
    print("No results found.")

display(data)
```
