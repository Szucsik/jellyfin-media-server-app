```python
start = time.time()

params = {
    "IncludeItemTypes": "Episode",
    "Recursive": "true",
    "Fields": "UserData,Id,Name,Type,ProductionYear,SeriesName,ParentIndexNumber,IndexNumber,MediaSources,Path",
    "Filters": "IsPlayed",
}


resp = requests.get(url, headers=headers, params=params, timeout=30)
resp.raise_for_status()
data = resp.json()

end = time.time()

print(f"Total runtime of the program is {end - start} seconds")
print("Total movies:", data["TotalRecordCount"])
display(data)
```

```python
episodes = []
start_index = 0

while True:
    params["StartIndex"] = start_index

    response = requests.get(url, params=params, headers=headers)
    response.raise_for_status()
    data = response.json()

    items = data.get("Items", [])
    if not items:
        break

    for ep in items:
        name = ep.get("Name")
        series = ep.get("SeriesName")
        season = ep.get("ParentIndexNumber")
        episode_num = ep.get("IndexNumber")
        path = ep.get("Path")

        episodes.append({
            "series": series,
            "season": season,
            "episode": episode_num,
            "name": name,
            "path": path
        })

    start_index += len(items)

# Print results
for ep in episodes:
    print(f"{ep['series']} - S{ep['season']:02}E{ep['episode']:02} - {ep['name']}")
    print(f"  Path: {ep['path']}")
```
